```python
import numpy as np
import matplotlib.pyplot as plt

def show_multiple_images_from_single_npy(file_path, H=480, W=640, clip_max=5.0):
    # 1. 读取数据
    data = np.load(file_path)  # shape: [N * H * W]
    total_pixels = data.shape[0]
    num_images = total_pixels // (H * W)

    print(f"📦 文件 {file_path} 中包含 {num_images} 张图像")

    for i in range(num_images):
        start = i * H * W
        end = (i + 1) * H * W
        depth_array = data[start:end]
        depth_image = depth_array.reshape(H, W)

        # 灰度映射
        depth_image = np.clip(depth_image, 0.0, clip_max)
        gray = (depth_image / clip_max * 255).astype(np.uint8)

        # 显示图像
        plt.figure(figsize=(8, 6))
        plt.imshow(gray, cmap='gray')
        plt.title(f"Image {i} from {file_path}")
        plt.colorbar(label="Depth (grayscale)")
        plt.axis('off')
        plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

def plot_lidar_pointcloud(file_path, index):
    data = np.load(file_path)
    if data.size % 3 != 0:
        print(f"{file_path} shape invalid")
        return
    pointcloud = data.reshape(-1, 3)

    fig = plt.figure(figsize=(10, 8))
    ax = fig.add_subplot(111, projection='3d')
    ax.scatter(pointcloud[:, 0], pointcloud[:, 1], pointcloud[:, 2],
               s=1, c=pointcloud[:, 2], cmap='viridis')
    ax.set_title(f"LIDAR Point Cloud (Frame {index})")
    ax.set_xlabel("X (m)")
    ax.set_ylabel("Y (m)")
    ax.set_zlabel("Z (m)")
    plt.tight_layout()
    plt.show()

# 遍历 lidar_0.npy 到 lidar_4.npy
for i in range(5):
    file = f"./lidar_{i}.npy"
    plot_lidar_pointcloud(file, i)
```
